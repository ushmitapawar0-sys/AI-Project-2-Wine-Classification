# Wine Classification: Exploratory Data Analysis

In this notebook, we explore the Wine Dataset from scikit-learn. We will look at class distribution, feature correlations, and distributions.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

%matplotlib inline
sns.set_theme(style="whitegrid")
```

## 1. Load Data

```python
data = load_wine()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
df['target_name'] = df['target'].map({0: data.target_names[0], 1: data.target_names[1], 2: data.target_names[2]})
df.head()
```

## 2. Class Balance
It is important to check if one class has significantly more samples than others. If the dataset is imbalanced, metrics like accuracy can be misleading.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='target_name')
plt.title("Class Distribution")
plt.show()
```

## 3. Correlation Heatmap
We use a heatmap to check for collinearity between features.

```python
plt.figure(figsize=(10, 8))
corr = df.drop(columns=['target', 'target_name']).corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
plt.title("Feature Correlation Heatmap")
plt.show()
```

## 4. Feature Distributions (Pairplot)
We plot a few key features to see how well they separate the classes. We pick the top 3 correlated features with the target for the pairplot.

```python
top_features = ['flavanoids', 'proline', 'od280/od315_of_diluted_wines', 'target_name']
sns.pairplot(df[top_features], hue='target_name', palette='Set2')
plt.suptitle("Pairplot of Top Features", y=1.02)
plt.show()
```
